```python
top_products = df.groupby("product_name")["sales"].sum()

top_products = top_products.sort_values(
    ascending=False
).head(5)

print(top_products)
```

```python
loss_products = df.groupby("product_name")["profit"].sum()

loss_products = loss_products.sort_values(
    ascending=True
)

loss_products = loss_products[loss_products < 0].head(5)

print(loss_products)
```

```python
region_sales = df.groupby("region")["sales"].sum()

region_sales = region_sales.sort_values(
    ascending=False
)

print(region_sales)
```
